# Indexing and Selection


Indexing means accessing specific rows or columns from a Pandas Series or DataFrame.

Selection means choosing particular rows, columns, or values from a dataset.

Pandas provides two important accessors:
1. `loc[]` – label-based indexing. It selects using row labels and column names.
2. `iloc[]` – position-based indexing. It selects using integer row and column positions.
We use indexing to retrieve a particular student's details, select specific columns, access individual values, or work with a subset of a dataset.

```python
import pandas as pd
import numpy as np
```

## 1. Creating a Sample DataFrame

```python
data={
    'Name':['Aman', 'Riya', 'Karan', 'Priya', 'Rahul'],
    'Age': [21, 22, 20, 23, 21],
    'City': ['Delhi', 'Agra', 'Pune', 'Delhi', 'Mumbai'],
    'Marks': [85, 90, 78, 95, 88]
}
df = pd.DataFrame(data, index=['a', 'b', 'c', 'd', 'e'])
print(df)
```

Observation: The row labels are a, b, c, d, and e. Column labels are Name, Age, City, and Marks.

## 2 Label-Based Indexing – loc[]

loc[] accesses rows and columns using their labels or names. It does not use their integer positions.

**df.loc[row_label, column_label]**

`row_label`: label of the row to select.

`column_label`: name of the column to select.

`:` means select all rows or all columns in that position.

### 2.1 Selecting a single row

Passing one row label returns that row as a Pandas Series.

```python
print(df.loc['a'])

print()
print("same output will be in both code") 
print()
print(df.loc['a',:])
```

### 2.2 Selecting a single value

Provide both a row label and a column name.

```python
print(df.loc['b','Name'])
```

### 2.3 Selecting a single column

Use : for all rows and provide the column name.

```python
print(df.loc[:,'Name'])
```

```python
print(df.loc['a',:])
```

### 2.4 Selecting multiple rows

Pass a list of row labels to select non-contiguous rows.

```python
print(df.loc[['a', 'c', 'e']])
```

### 2.5 Selecting multiple columns

Pass a list of column names. This returns a DataFrame.

```python
print(df.loc[:,['Name','Marks']])
```

### 2.6 Selecting specific rows and columns

```python
print(df.loc[['a','c'],['Name','Marks']])
```

### 2.7 Selecting rows using label slicing

```python
print(df.loc['a':'d'])
# a and d both include
```

## 3. Position-Based Indexing – iloc[]

iloc[] is used to access rows and columns using their integer positions.

It does not use custom index labels or column names. It uses positions starting from 0.

**df.iloc[row_position, column_position]**

`row_position`: Integer position of the row.

`column_position`: Integer position of the column.


### 3.1 Selecting a single row

```python
print(df.iloc[0])
# This selects the first row because its position is 0.
```

### 3.2 Selecting a single value

```python
print(df.iloc[1,0])

# 1 represents the second row.
# 0 represents the first column.
```

### 3.3 Selecting a single column

```python
print(df.iloc[:,0])
```

### 3.4 Selecting multiple rows

```python
print(df.iloc[[0,2,4]])

#This selects rows at positions 0, 2, and 4.
```

### 3.5 Selecting multiple columns

```python
print(df.iloc[:,[0,3]])
# This selects all rows from the first and fourth columns (Name and Marks).
```

### 3.6 Selecting specific rows and columns

```python
print(df.iloc[[0,2],[0,3]])
#This selects:
# - Rows at positions 0 and 2.
# - Columns at positions 0 and 3.
```

###  3.7 Selecting rows using position slicing

```python
print(df.iloc[0:3])
```

```python
print(df.iloc[0:3])


# This selects rows at positions 0, 1, and 2.
##Important: Position-based slicing with iloc[] excludes the ending position.
### Therefore, df.iloc[0:3] selects positions 0, 1, and 2—not 3.
```

### 4. Label-Based vs Position-Based Indexing

Feature|	loc[]|	iloc[]|
--|--|--|
Indexing type|	Label-based|	Position-based
Uses|	Row labels and column names|	Integer positions
Row selection|	df.loc['a']|	df.iloc[0]
Column selection|	df.loc[:, 'Name']	|df.iloc[:, 0]
Slicing end|	Included|	Excluded
Custom index|	Uses actual labels|	Uses integer positions

```python
# Label-based indexing
print(df.loc['b', 'Marks'])

# Position-based indexing
print(df.iloc[1, 3])
```

### Remember:
`loc[]` → Think of labels.

`iloc[]` → Think of integer positions.

### 5. Selecting Multiple Rows and Columns

Sometimes we need to select a combination of rows and columns rather than just one row or one column.

Both loc[] and iloc[] support selecting multiple rows and columns.

### 5.1 Using loc[]

```python
print(df.loc[['a','b','d'],['Name','City']])

# Rows with labels a, b, and d.
# Columns Name and City.
```

### 5.2 using iloc[]

```python
print(df.iloc[[0, 1, 3], [0, 2]])


#- Rows at positions 0, 1, and 3.
#- Columns at positions 0 and 2.
```

### 5.3 Selecting a range of rows and columns using loc[]

```python
print(df.loc['a':'d', 'Name':'City'])

# This selects rows from label a through d and columns from Name through City. Both endpoints are included.
```

### 5.4 Selecting a range of rows and columns using iloc[]

```python
print(df.iloc[0:4, 0:3])

# This selects rows at positions 0 through 3 and columns at positions 0 through 2. The ending positions are excluded.
```

### 6. Selecting Rows Based on Conditions

Although conditional filtering is often covered in a separate Pandas topic, it's useful to understand how conditions work with loc[].

```python
# Example 1: Selecting students with marks greater than 85
print(df.loc[df['Marks'] > 85])
```

```python
# Example 2: Selecting students from Delhi

print(df.loc[df['City'] == 'Delhi'])
```

```python
# Example 3: Selecting specific columns for matching rows
print(df.loc[df['Marks'] > 85, ['Name', 'Marks']])
```

### Important: 
Boolean conditions can be passed directly into loc[] to filter rows.

## Important Points to Remember
1.loc[] works with labels; iloc[] works with integer positions.

2.loc[] includes the ending label when slicing.

3.iloc[] excludes the ending position when slicing.

4.Use a list inside the accessor to select multiple non-contiguous rows or columns.

5.Use : to select all rows or all columns.

6. A single row or column selection generally returns a Series; selecting multiple columns using a list returns a DataFrame.

7. Invalid labels or out-of-range integer positions raise an error.
