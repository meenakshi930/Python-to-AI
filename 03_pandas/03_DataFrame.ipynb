# DataFrame

A **DataFrame** is a two-dimensional labelled data structure, like a table. Rows usually represent records and columns represent variables or features.

```python
import pandas as pd
import numpy as np
```

## 1. Creating DataFrames

use `pd.DataFrame(data, index=None, coulmns=None, dtype=None)`

| Argument | Purpose |
|---|---|
| `data` | Input data |
| `index` | Optional row labels |
| `columns` | Optional column labels |
| `dtype` | Optional data type |

A DataFrame has rows (axis 0) and columns (axis 1).

```python
data = {'state': ['Ohio', 'Ohio', 'Ohio', 'Nevada', 'Nevada'],
        'year': [2000, 2001, 2002, 2001, 2002],
        'pop': [1.5, 1.7, 3.6, 2.4, 2.9]}
frame = pd.DataFrame(data)
print(frame)
```

## 2. DataFrame from dictionaries

### Dictionary of lists

Dictionary keys become column names; list items become values in those columns. All lists must have the same length.

### Dictionary of dictionaries

Outer keys become column names. Inner dictionary keys become row labels.

### Custom row index

Pass `index=` when creating the DataFrame. The number of labels must match the number of rows.

```python
data = {
    'state': ['Ohio', 'Ohio', 'Ohio', 'Nevada', 'Nevada'],
    'year': [2000, 2001, 2002, 2001, 2002],
    'pop': [1.5, 1.7, 3.6, 2.4, 2.9]
}

frame = pd.DataFrame(data)
print(frame)
```

## 3. DataFrame from lists

### List of lists

Each inner list represents one row. Provide `columns=` to name the columns.

### List of dictionaries

Each dictionary represents one row. Dictionary keys become column names. This format is common when data comes from JSON APIs.

### From a NumPy array

A two-dimensional NumPy array can also be used to create a DataFrame.

### Compare input formats

| Input | Interpretation |
|---|---|
| Dictionary of lists | Keys are columns; list items are values |
| Dictionary of dictionaries | Outer keys are columns; inner keys are row labels |
| List of lists | Each inner list is a row |
| List of dictionaries | Each dictionary is a row |
| 2D NumPy array | Each array row becomes a DataFrame row |

```python
data = [
    ['Aman', 21, 'Delhi'],
    ['Riya', 22, 'Mumbai'],
    ['Karan', 20, 'Pune']
]

frame = pd.DataFrame(data, columns=['Name', 'Age', 'City'])
print(frame)
```

## 4. DataFrame attributes

Attributes provide information and are accessed **without parentheses**.

| Attribute | Meaning |
|---|---|
| `shape` | Tuple: (rows, columns) |
| `size` | Total number of cells |
| `columns` | Column labels |
| `index` | Row labels |
| `ndim` | Number of dimensions |
| `dtypes` | Data type of each column |

```python
print(frame.shape)
print(frame.size)
print(frame.columns)
print(frame.index)
```

## 5. `head()` and `tail()`

- `head(n)` returns the first `n` rows (default: 5).
- `tail(n)` returns the last `n` rows (default: 5).

Both are methods, so use parentheses. They are useful for a quick preview of an unfamiliar dataset.

```python
print(frame.head())

print(frame.head(2))
```

```python
print(frame.tail())
print(frame.tail(2))
```

## 6. `info()` and `describe()`

### `info()`

Prints a concise summary: index, row and column counts, column names, non-null counts, data types, and memory usage. It is useful for spotting missing values and checking data types.

`info()` prints its output and normally returns `None`.

### `describe()`

Returns descriptive statistics. By default, it summarises numeric columns.

| Statistic | Meaning |
|---|---|
| `count` | Number of non-missing values |
| `mean` | Average |
| `std` | Standard deviation |
| `min`, `max` | Smallest and largest |
| `25%`, `50%`, `75%` | Quartiles; 50% is the median |

Use `describe(include="all")` to include non-numeric columns. The statistics shown depend on the column data type.

```python
print(frame.info())
```

```python
print(frame.describe())
```

## 7. Selecting columns

- `df["Column"]` selects one column and returns a **Series**.
- `df[["Column1", "Column2"]]` selects multiple columns and returns a **DataFrame**.
- Prefer square-bracket notation over dot notation: it works with spaces and names that conflict with DataFrame attributes or methods.

You can also use `.loc[row_selection, column_selection]`; `:` means all rows.

```python
print(frame['Name'])

print(frame[['Name', 'Age']])
```

## 8. Adding columns

Common ways to add columns:

1. Assign a list: `df["City"] = ["Delhi", "Agra", ...]` (length must match the number of rows).
2. Assign a scalar: `df["Country"] = "India"` (broadcasts the value to every row).
3. Calculate from existing columns: `df["Annual"] = df["Monthly"] * 12`.
4. Use a condition: `np.where(condition, value_if_true, value_if_false)`.
5. Insert at a specific position: `df.insert(position, "NewColumn", values)`.

Assignment and `insert()` modify the DataFrame.

```python
frame['Marks'] = [85, 90, 78]

print(frame)
```

```python
frame['Age'] = frame['Age'] + 1

print(frame)
```

## 9. Deleting columns

| Method | Effect |
|---|---|
| `df.drop(columns=["A"])` | Returns a new DataFrame by default |
| `del df["A"]` | Deletes directly from the original DataFrame |
| `df.pop("A")` | Deletes from original and returns the removed column as a Series |

Prefer `drop()` when you want to keep the original DataFrame unchanged. `del` and `pop()` modify it.

```python
frame = frame.drop('City', axis=1)

print(frame)
```

```python
# 8. Select only `Name`; check its type.
print(frame['Name'])
```

```python
# 9. Select `Name` and `Marks`; check its type.
print(frame[['Name','Marks']])
```

```python
# 10. Add a Pass/Fail column (pass if marks are at least 40).
frame['Result'] = np.where(frame['Marks'] >= 40, 'Pass', 'Fail')

print(frame)
```

```python
# 11. Add 5 to every student's marks in a new column.
frame['Marks_plus_5']=frame['Marks']+5
print(frame)
```

```python
# 12. Drop `City` without changing the original DataFrame.
new_frame = frame.drop(columns=['City'])

print(new_frame)
print(frame)  # Original DataFrame remains unchanged
```

```python
# 13. Remove a column using `del` or `pop()` and observe the difference.
import pandas as pd

data = {
    'Name': ['Mee', 'Pri', 'Kri', 'Ani'],
    'Age': [14, 15, 17, 21],
    'Marks': [99, 54, 78, 95],
    'City': ['Goa', 'UP', 'UK', 'Sikkim']
}

frame = pd.DataFrame(data)

# Using del
del frame['City']

print("After using del:")
print(frame)

# Using pop()
removed_column = frame.pop('Marks')

print("\nRemoved column:")
print(removed_column)

print("\nAfter using pop():")
print(frame)
```

## 11. Quick revision

| Concept | Syntax |
|---|---|
| Create DataFrame | `pd.DataFrame(data)` |
| From dictionary | `pd.DataFrame({"A": [1, 2]})` |

| Total cells | `df.size` |

| Row labels | `df.index` |

| Select one column | `df["A"]` |
| Select multiple columns | `df[["A", "B"]]` |
| Add a column | `df["New"] = values` |
| Insert a column | `df.insert(position, "New", values)` |
| Drop a column | `df.drop(columns=["A"])` |
| Delete a column | `del df["A"]` or `df.pop("A")` |
