# 02. Pandas Series
A **Series** is a one-dimensional labelled array. Each value has an associated index label.

````python
import numpy as np
import pandas as pd
````

# 1. Creating a Series

Use `pd.Series()` to create a Series.

```python
pd.Series(data, index=None, dtype=None, name=None)
```

| Parameter | Meaning |
|---|---|
| `data` | Values: list, dictionary, array, scalar, etc. |
| `index` | Labels associated with values |
| `dtype` | Optional data type |
| `name` | Optional name for the Series |

If no index is supplied, Pandas creates a default integer index starting at zero.

````python
marks =pd.Series([85,92,78,54])
print(marks)
print(type(marks))
````

The left side is the index and the right side contains the values. For example, label 0 refers to 85.

A Series has both values and labels. Those labels matter when selecting data and performing operations.

# 2. Series from lists

A Python list is a common input to `pd.Series()`.

````python
# Integer list
temperatures = pd.Series([29, 31, 30, 28, 32])
temperatures
````

````python
# String list
cities = pd.Series(["Delhi", "Agra", "Jaipur", "Mathura"])
cities
````

````python
# Floating-point list
prices = pd.Series([99.5, 149.0, 79.99, 210.5])
prices
````

````python
# Specify a data type
scores = pd.Series([10, 20, 30, 40], dtype="int64")
print(scores)
print(scores.dtype)
````

````python
# Give a Series a name
monthly_sales = pd.Series([12000, 15000, 18000], name="Monthly Sales", dtype="int32")
monthly_sales
````

The `dtype` argument requests a data type. Pandas will raise an error if the values cannot be converted to that type.

# 3. Series from dictionaries

A dictionary stores key-value pairs. When converted to a Series, its keys become index labels and its values become Series values.

````python
students_marks={
    "Asha":85,
    "Riya":95,
    "Meenakshi":99,
    "Kabir": 88
}
marks=pd.Series(students_marks)
marks
````

````python
# Access dictionary-based Series values
print(marks["Riya"])
print(marks["Kabir"])
````

````python
# Reorder using a requested index
# When an index is supplied, Pandas arranges values in that label order. A requested label missing from the dictionary receives a missing value.
ordered_marks = pd.Series(
    students_marks,
    index=["Meera", "Asha", "Riya", "Arjun"]
)
ordered_marks

# Arjun is not a key in the dictionary, so its value is missing (NaN in this example).
````

# 4. Custom indexes

A custom index replaces the default 0, 1, 2, ... labels with meaningful labels.

````python
marks_by_id = pd.Series(
    [85, 92, 78, 88],
    index=["S101", "S102", "S103", "S104"]
)
marks_by_id
````

````python
# Access a value by its custom label
marks_by_id.loc["S102"]
````

````python
# Change index labels
marks_by_id.index=["ST01", "ST02", "ST03", "ST04"]
marks_by_id
````

### 4.1 Set the index during creation or afterwards

````python
# Approach 1: Set the index during creation
series_a = pd.Series([10, 20, 30], index=["x", "y", "z"])

# Approach 2: Assign the index afterwards
series_b = pd.Series([10, 20, 30])
series_b.index = ["x", "y", "z"]

print(series_a)
print(series_b)
````

# 5. Indexing a Series

Indexing means selecting one or more values.

Two important accessors:

- loc selects by **index label**.
- iloc selects by **integer position**.

````python
scores = pd.Series(
    [85, 92, 78, 88, 95],
    index=["Asha", "Riya", "Kabir", "Meera", "Arjun"]
)
scores
````

````python
# Label-based indexing with loc
print(scores.loc["Riya"])
print(scores.loc["Arjun"])
````

````python
# Position-based indexing with iloc
print(scores.iloc[0])    # First value
print(scores.iloc[4])    # Last value
print(scores.iloc[-1])  # Last value
````

````python
# Select multiple labels
scores.loc[["Asha", "Meera", "Arjun"]]
````

````python
# Select multiple positions
scores.iloc[[0, 2, 4]]
````

````python
# Select values using a condition
scores[scores >= 90]
````

# 6. Slicing a Series

````python
# Position-based slicing with iloc
scores.iloc[1:4]
# iloc follows Python slicing: start is included and stop is excluded.
# This selects positions 1, 2, and 3, but not position 4.
````

````python
scores.iloc[:3]
#  First three values
````

````python
scores.iloc[2:]
#  From position 2 to the end
````

````python
scores.iloc[::-1]  # Reverse the Series
````

````python
scores.loc["Asha": "Meera"]
````

This includes both `Asha` and `Meera`.

| Accessor | Selects by | Slice stop |
|---|---|---|
| `iloc` | Integer position | Excluded |
| `loc` | Index label | Included |

Use `loc` when you mean labels and `iloc` when you mean positions.

# 7. Series attributes

Attributes provide information about a Series and are accessed without parentheses.

````python
sales=pd.Series(
    [1200,1500,1100,1800],
    index=["Monday","Tuesday","Wednesday","Thursday"],
    name="daily Sales"
)
sales
````

| Attribute | Meaning |
|---|---|
| `values` | Underlying values |
| `index` | Index labels |
| `dtype` | Data type of values |
| `size` | Number of elements |
| `shape` | Tuple containing dimensions |
| `ndim` | Number of dimensions |
| `name` | Series name |

````python
print("Values:", sales.values)
print("Index:", sales.index)
print("Data type:", sales.dtype)
print("Size:", sales.size)
print("Shape:", sales.shape)
print("Dimensions:", sales.ndim)
print("Name:", sales.name)
````

### Attributes vs methods

- `sales.dtype` is an attribute: no parentheses.
- `sales.head()` is a method: parentheses.
- `sales.sum()` is a method.
- `sales.mean()` is a method.

Useful methods:

````python
print("First two values:")
print(sales.head(2))
print("Total:",sales.sum())
print("Average:",sales.mean())
print("Minimum:",sales.min())
print("maximum:",sales.max())
print("Non-missing count:",sales.count())
````

# 8. Vectorized operations

**Vectorization** means applying an operation to an entire Series without writing an explicit Python loop for each element. It makes data manipulation concise and often efficient.

````python
prices = pd.Series(
    [100, 200, 300, 400],
    index=["Pen", "Notebook", "Bag", "Bottle"]
)
prices
````

### 8.1 Arithmetic operations

````python
prices+10
````

````python
prices-20
````

````python
prices*2
````

````python
prices/2
````

````python
prices**2
````

### 8.2 Comparison operations

Comparisons return Boolean values (`True` or `False`) for each element.

````python
prices>200
````

````python
prices==200
````

````python
prices<=200
````

### 8.3 Boolean filtering

````python
prices[prices > 200]
````

````python
prices[(prices >= 100) & (prices <= 300)]
````

For multiple conditions:

- `&` means AND.
- `|` means OR.
- `~` means NOT.
- Put each condition inside parentheses.

Do not use Python's `and` or `or` directly between Series conditions.

###  8.4 Common mathematical methods

````python
numbers = pd.Series([10, 20, 30, 40, 50])

print("Sum:", numbers.sum())
print("Mean:", numbers.mean())
print("Median:", numbers.median())
print("Minimum:", numbers.min())
print("Maximum:", numbers.max())
print("Standard deviation:", numbers.std())
````

### 8.5 Apply a NumPy function

````python
np.sqrt(numbers)
````

### 8.6 Operations between two Series: index alignment

A key Pandas behaviour: arithmetic between Series aligns values by **index label**, not simply by position.

````python
first = pd.Series([10, 20, 30], index=["A", "B", "C"])
second = pd.Series([1, 2, 3], index=["B", "C", "D"])

first + second
````

Explanation:

- Label `B`: 20 + 1 = 21.
- Label `C`: 30 + 2 = 32.
- Label `A` has no match in `second`.
- Label `D` has no match in `first`.

Unmatched labels produce missing results (`NaN` in this example). This is called **index alignment**. It is useful when combining data from different sources, but you should understand it to avoid unexpected missing values.

### Use `fill_value` with arithmetic methods

````python
first.add(second, fill_value=0)
````

`fill_value=0` treats a missing counterpart as zero for this addition. Only use zero when it is meaningful for your data; it is not always an appropriate replacement.

# 9. Missing values in a Series

A Series can contain missing values. Pandas commonly displays missing numeric values as `NaN`.

````python
attendance = pd.Series(
    [95, 88, np.nan, 92, np.nan],
    index=["Asha", "Riya", "Kabir", "Meera", "Arjun"]
)
attendance
````

### Detect and count missing values

````python
print(attendance.isna())
print("Missing count:", attendance.isna().sum())
print("Non-missing count:", attendance.count())
````

### Remove or replace missing values

````python
attendance.dropna()
````

````python
attendance.fillna(0)
````

`dropna()` and `fillna()` return a result; they do not normally overwrite the original Series unless you assign the result back.
