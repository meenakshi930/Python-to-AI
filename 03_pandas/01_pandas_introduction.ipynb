# Getting Started with Pandas

Pandas is an open-source Python library that provides high-level data structures and tools for analysing data. Its two central data structures are:
- Series: a one-dimensional labelled array.
- DataFrame: a two-dimensional labelled table with rows and columns.
Pandas is especially useful when data is arranged in rows and columns, such as CSV files, Excel sheets, database tables, and many datasets used in data science.

A simple example
Imagine that you have student data:
Name	Age	Marks
Asha	20	85
Riya	21	92
Kabir	19	78

Pandas lets you load this data into a DataFrame and then ask questions such as:
- What is the average mark?
- Which students scored more than 80?
- Are any values missing?
- How many students are in the dataset?
- Which columns should be used as model features?

Where does Pandas fit in an AI/ML workflow? A simplified machine-learning workflow looks like this:

1. Collect data (CSV, Excel, database, API, etc.).
2. Load the data into Pandas.
3. Inspect and understand the data.
4. Clean missing, duplicate, or inconsistent values.
5. Transform and prepare features.
6. Pass suitable data to a machine-learning library such as scikit-learn.
7. Train and evaluate a model.


Pandas is commonly used in steps 2–5. It does not replace machine-learning libraries; it helps prepare and analyse the data those libraries use.

# Why Pandas require?

Raw data is rarely ready for analysis. It may contain missing values, duplicate records, inconsistent text, incorrect data types, or columns that are not useful.

Pandas provides convenient operations for common data tasks.

_Task:-What Pandas helps you do_ 

* Load data-Read CSV, Excel, JSON, SQL query results, and more
* Inspect data-View rows, columns, data types, and summary information
* Select data-Select particular rows, columns, or conditions
* Clean data-Handle missing values and duplicates
* Transform data-Create columns, rename columns, change data types
* Analyse data-Calculate counts, averages, minimums, maximums, and group summaries
* Combine data-Merge, join, and concatenate tables
* Prepare ML data-Select features, filter records, and prepare tabular datasets

Why not use only Python lists and dictionaries?
You can work with data using built-in Python structures, but Pandas provides table-oriented operations that are more convenient for many data-analysis tasks. For example, calculating a column average or filtering rows by a condition is straightforward.
Pandas and NumPy
Pandas is built on top of and works closely with NumPy.
- NumPy is especially useful for numerical arrays and mathematical operations.
- Pandas adds labels, tabular structures, flexible indexing, and data-analysis tools.
You will often use both in data science and machine learning.

# Series vs DataFrame

4.1 Series

A Series is a one-dimensional labelled array. It contains:
- A sequence of values.
- An index that labels each value.

A Series can hold integers, floating-point numbers, strings, Python objects, and other supported data types.

```python
marks = pd.Series([85, 92, 78, 88])

print(marks)
```

```python
# Create a Series with a custom index
marks = pd.Series(
    [85, 92, 78],
    index=["Asha", "Riya", "Kabir"]
)

print(marks)
```

```python
# Access values in a Series
print(marks["Riya"])  # Access by index label
print(marks.iloc[0])  # Access by integer position

# .loc[...] selects by index label.
# .iloc[...] selects by integer position.
```

4.2 DataFrame


A DataFrame is a 2-dimensional labelled data structure. You can think of it as a table or spreadsheet:
- Each row represents a record.
- Each column represents a variable or feature.
- Columns can have different data types.
- Rows and columns have labels.

Each column in a DataFrame is a Series

```python
#Create a DataFrame from a dictionary
student_data = {
    "Name": ["Asha", "Riya", "Kabir", "Meera"],
    "Age": [20, 21, 19, 20],
    "Marks": [85, 92, 78, 88],
    "City": ["Delhi", "Jaipur", "Agra", "Mathura"]
}

students = pd.DataFrame(student_data)

students
```

```python
# Create a DataFrame with a custom index
students_indexed = pd.DataFrame(
    student_data,
    index=["S01","S02","S03","S04"]
)
students_indexed
```

4.3 Selecting a column: DataFrame to Series

```python
marks_column=students["Marks"]
print(marks_column)
print(type(marks_column))
```

```python
one_column_dataframe=students[["Marks"]]
print(type(one_column_dataframe))
one_column_dataframe
```
